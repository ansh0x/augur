The species gap between train and test/val split is very huge. So, I'm gonna make a stratified 20% split of the species that are in training data but not in val/test split and test on them at the end of the model trainng, by combing it with the test data.

# Mapping the Test/Val 

```python
id_to_name = annot_df.set_index('id_cat')['name_id'].to_dict()
id_to_super = annot_df.set_index('id_cat')['supercategory_id'].to_dict()
```

```python
val_filtered_annot['name_id'] = val_filtered_annot['category_id'].map(id_to_name)
val_filtered_annot['supercategory_id'] = val_filtered_annot['category_id'].map(id_to_super)

test_filtered_annot['name_id'] = test_filtered_annot['category_id'].map(id_to_name)
test_filtered_annot['supercategory_id'] = test_filtered_annot['category_id'].map(id_to_super)
```

```python
val_filtered_annot.to_csv('val_annotation.csv', index=False)
```

```python
from sklearn.model_selection import StratifiedKFold, train_test_split
```

```python
not_in_val_df = annot_df[~annot_df['id_cat'].isin(test_filtered_annot['category_id'])]
train, test = train_test_split(not_in_val_df, stratify=not_in_val_df['id_cat'], shuffle=True, random_state=42)
```

```python
test.to_csv('rare_test.csv', index=True)
```

```python
annot_df.drop(test.index, inplace=True)
```

```python
annot_df.to_csv('augmented_annot.csv')
```

```python
skf = StratifiedKFold(n_splits=3, shuffle=True)
```

```python
test_x = test_filtered_annot.copy()
test_y = test_x.pop('name_id')
```

```python
for i, (_, test_idx) in enumerate(skf.split(test_x, test_y)):
    fold = test_filtered_annot.iloc[test_idx].reset_index(drop=True)
    fold.to_csv(f'test_split_{i}.csv', index=False)
```
