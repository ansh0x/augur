* As it can be seen that sample count is very skewed. Thus, I'll be a doing a mix of oversampling and undersampling to keep species sample counts balanced.

* I'll be doing it supercatergory wise to becasue of my hierarchical model structure.

```python
def sample_data(data):
    groups = data.groupby('name')
    aug_type = ['Pitch Shift', 'Time Shift', 'Time Stretch', 'Noise', 'Gain']
    new_data = data
    temp = []
    for name, group in groups:
        if len(group) >= 500:
            num_to_drop = len(group) - 300 # I'll only keep 300 samples for a species at most
            idx_to_drop = group.sample(num_to_drop).index
            new_data.drop(index=idx_to_drop, inplace=True)
        elif len(group) > 10 and len(group) <= 50: # I'll oversample species with samples between 11-50, 5 times of each sample 
            for aug in aug_type:
                g = group.copy()
                g['aug_type'] = aug
                temp.append(g)
        elif len(group) <= 10: # I'll oversample species with samples <10 , 10 times of each sample 
            for aug in aug_type:
                g = group.copy()
                g['aug_type'] = aug
                temp.append(g)
                temp.append(g)
    new_data = pd.concat([new_data, pd.concat(temp)])
    return new_data
```

```python
sampled_df = []
for name, group in group_super_cat:
    if name in ['Amphibia', 'Insecta']: continue # I'm not using Amphibia and Insecta species because of the project scope.
    sampled_df.append(sample_data(group))
df = pd.concat(sampled_df)
```
